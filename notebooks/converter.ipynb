```python
import os
import glob

import numpy as np
import rasterio
from PIL import Image


def normalizar_banda(banda):
    banda = banda.astype(np.float32)

    minimo = np.nanpercentile(banda, 2)
    maximo = np.nanpercentile(banda, 98)

    if maximo <= minimo:
        return np.zeros_like(banda, dtype=np.uint8)

    banda = np.clip(banda, minimo, maximo)
    banda = (banda - minimo) / (maximo - minimo)

    return (banda * 255).astype(np.uint8)


pasta_tif = "../data/raw/images"
pasta_png = "../data/roboflow/images"

os.makedirs(pasta_png, exist_ok=True)

arquivos = sorted(glob.glob(f"{pasta_tif}/*.tif"))

for caminho in arquivos[:60]:
    id_imagem = os.path.splitext(
        os.path.basename(caminho)
    )[0]

    with rasterio.open(caminho) as src:
        b02 = src.read(1)
        b03 = src.read(2)
        b04 = src.read(3)

    rgb = np.dstack([
        normalizar_banda(b04),
        normalizar_banda(b03),
        normalizar_banda(b02)
    ])

    saida = os.path.join(
        pasta_png,
        f"{id_imagem}.png"
    )

    Image.fromarray(rgb).save(saida)

print("Conversão concluída.")
```
